```python
# load json file
import json
import numpy as np
import pandas as pd 
from tqdm import tqdm 

def load_json_file(file_path):
    with open(file_path, 'r') as f:
        json_data = json.load(f)
    # print('Total number of items in the json file:', len(json_data))
    return json_data
```

```python
def edit_distance(str1, str2):
    m = len(str1)
    n = len(str2)
    
    dp = np.zeros((m + 1, n + 1), dtype=int)

    for i in range(m + 1):
        for j in range(n + 1):
            if i == 0:
                dp[i][j] = j
            elif j == 0:
                dp[i][j] = i
            elif str1[i - 1] == str2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                dp[i][j] = 1 + min(dp[i - 1][j],    # Deletion
                                   dp[i][j - 1],    # Insertion
                                   dp[i - 1][j - 1]) # Substitution
    return dp[m][n]
# edit_distance(json_data)
```
