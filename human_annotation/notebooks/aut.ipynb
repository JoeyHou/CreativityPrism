```python
aut_eval_dir = '../data/raw/aut/evaluation_outputs/'
import os
import pandas as pd

all_tools = []
responses = []
for dir in os.listdir(aut_eval_dir):
    if dir == '.DS_Store': continue
    for csv_file in os.listdir(aut_eval_dir + dir):
        if '.csv' not in csv_file: continue
        tmp_df = pd.read_csv(aut_eval_dir + dir + '/' + csv_file)
        tmp_tools = tmp_df['prompt'].unique()
        for t in tmp_tools:
            if t not in all_tools:
                all_tools.append(t)
        if 'response' in tmp_df.columns:
            responses.extend(list(zip(
                tmp_df['prompt'].values,
                tmp_df['response'].values,
                tmp_df['target'].values,
            )))
```

```python
i = 0
found, not_found = 0, 0
bottle_demos = [j for j, score in aut_demos['bottle']]
# while responses[i][0] == 'bottle':
non_demo_human_annotation = []
all_response_text = []
for key, response, score in responses:
    score = float(score)
    if key != 'bottle':
        continue
    if response in bottle_demos:
        found += 1
    else:
        if response not in all_response_text:
            all_response_text.append(response)
            non_demo_human_annotation.append((key, response, score))
        not_found += 1
    # i += 1
print(found, not_found, len(non_demo_human_annotation))
```
