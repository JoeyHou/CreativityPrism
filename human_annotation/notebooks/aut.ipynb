# 3. Load All respnses 

```python
aut_eval_dir = '../data/raw/aut/evaluation_outputs/'
import os
import pandas as pd

all_tools = []
human_responses = []
for dir in os.listdir(aut_eval_dir):
    if dir == '.DS_Store': continue
    for csv_file in os.listdir(aut_eval_dir + dir):
        if '.csv' not in csv_file: continue
        tmp_df = pd.read_csv(aut_eval_dir + dir + '/' + csv_file)
        tmp_tools = tmp_df['prompt'].unique()
        for t in tmp_tools:
            if t not in all_tools:
                all_tools.append(t)
        if 'response' in tmp_df.columns:
            human_responses.extend(list(zip(
                tmp_df['prompt'].values,
                tmp_df['response'].values,
                tmp_df['target'].values,
            )))
```

# 4. Sample from Responses

```python
random.seed(42)

response_per_group = 40

groupped_responses = {}
for group, response, score in human_responses:
    if group not in groupped_responses:
        groupped_responses[group] = []
    if response in [res for res, _ in aut_demos[group]]:
        continue
    groupped_responses[group].append((response, score))

for group in groupped_responses:
    random.shuffle(groupped_responses[group])
    groupped_responses[group] = [[res, float(score)] for res, score in groupped_responses[group][:response_per_group]]

print({group: len(responses) for group, responses in groupped_responses.items()})
# groupped_responses
```

```python
subset_tool_score_all = []
for tool in groupped_responses:
    for use, score in groupped_responses[tool]:
        subset_tool_score_all.append({
            'tool': tool,
            'use': use,
            'score': score
        })
        for tmp_tool, tmp_use, tmp_score in human_responses:
            if tool == tmp_tool and use == tmp_use:
                if score != tmp_score:
                    subset_tool_score_all.append({
                        'tool': tool,
                        'use': use,
                        'score': tmp_score
                    })
subset_tool_score_df = pd.DataFrame(subset_tool_score_all)
subset_tool_score_df.shape
```

```python
def format_demo(tool):
    demos = aut_demos[tool]
    all_demo_text = ''
    for demo_text, demo_score in demos:
        # print(demo_text, demo_score)
        all_demo_text += f"\t{demo_text} (Score: {demo_score})<br>"
    return all_demo_text
# subset_tool_score_df['demo'] = subset_tool_score_df['tool'].apply(lambda tool: format_demo(tool))
# subset_tool_score_df
```

```python
# subset_tool_score_df.score.describe()'
import copy
```

```python
mturk_data = []
use_per_tool = 5
for tool in subset_tool_score_df['tool'].unique():
    # print(tool)
    # print(subset_tool_score_df[subset_tool_score_df['tool'] == tool]['score'].describe())
    # print()
    tmp_df = copy.deepcopy(subset_tool_score_df[subset_tool_score_df['tool'] == tool])
    tmp_mturk_data = {'tool_{}'.format(i+1): tmp_df['use'].values[i] for i in range(use_per_tool)}
    tmp_mturk_data['demo_text'] = format_demo(tool)
    tmp_mturk_data['tool_name'] = tool
    mturk_data.append(tmp_mturk_data)
mturk_df = pd.DataFrame(mturk_data)
mturk_df
```

## 4.2 Create Input for CreativityPrism

```python
# create input data 

all_input = []
for tool in groupped_responses:
    # tmp_input = []
    for i, (response, score) in enumerate(groupped_responses[tool]):
        all_input.append({
            "meta_data": {
                "dataset": "aut_push",
                "id": "{}-{}".format(tool, i),
                "eval_func": None,
                "human_score": score
            },
            "input": {
                "text": "Create a list of creative alternative uses for a bottle. They should be 5 words long. No adjectives.",
                "file": "",
                "others": {
                    "object": tool,
                    "prompt_type": "nc",
                    "iteration_lst": {}
                }
            },
            "output": {
                "content": response
            }
        })
with open(f'../data/aut_llmj_eval_fake_input.json', 'w') as f:
    json.dump(all_input, f, indent=4)

all_input[:3]
```
