```python
import copy 
import os
import pandas as pd
import random
random.seed(42)
# Directory containing neocoder output files
data_dir = "../data/raw/neocoder"
```

```python
# List all CSV files in the directory
# files = [f for f in os.listdir(data_dir) if f.endswith('.csv')]

# files
model_lst = [
    'Qwen2.5-7B-Instruct',
    'gpt-4.1',
    'OLMo-2-1124-13B-Instruct'
]
files = [mo + '_creativity.csv' for mo in model_lst]

rows_per_model = 20
min_rows = min([pd.read_csv(os.path.join(data_dir, file)).shape[0] for file in files])

dfs = []
for file in files:
    file_path = os.path.join(data_dir, file)
    df = pd.read_csv(file_path).dropna(subset=['machine_solutions'])
    # Add a column for model name (derived from file name, without extension)
    model_name = os.path.splitext(file)[0]
    df['model'] = model_name
    random.seed(42)
    randomly_sampled_rows = random.sample(range(min(min_rows, df.shape[0])), rows_per_model)
    df = copy.deepcopy(df.iloc[randomly_sampled_rows])
    dfs.append(df)

# Concatenate all dataframes into one big table
big_table = pd.concat(dfs, ignore_index=True)
```
